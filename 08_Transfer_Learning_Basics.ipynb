# 8 Transfer Learning Basics
The idea of transfer learning in computer vision. The main idea is simple: instead of training a CNN from scratch, we reuse a model that was already trained on a large dataset, such as ImageNet, and adapt it to a new task.

Transfer learning is very useful because learning visual features from scratch requires large datasets, more time, and more computation. A pretrained model already knows basic patterns such as edges, corners, textures, colors, and object parts.


## 1. What is Transfer Learning?

Transfer learning means using knowledge learned on one task and applying it to a different but related task. For example, a model trained on ImageNet can learn general visual features and then be reused for a medical image or flower classification problem.

This works because early layers in CNNs capture general patterns, while deeper layers capture task-specific patterns.


## 2. Why Pretrained Models are Useful

Pretrained models are important because:
- they save training time and compute
- they work well with small datasets
- they usually produce better performance than training from scratch
- they reduce the need for large labeled datasets

A pretrained model already contains useful feature extractors, so the new task only needs to adapt the final decision layer or a few upper layers.


## 3. Feature Extraction

In feature extraction, the backbone of a pretrained model is used as a fixed feature generator. The convolutional layers are frozen, meaning their weights are not updated. Only a new classifier, usually one or two fully connected layers, is trained for the new task.

This is a practical and fast approach when the new dataset is small.


## 4. Frozen Layers vs Trainable Layers

- Frozen layers: weights are fixed and do not change during training.
- Trainable layers: weights are updated according to the new dataset.

When the backbone is frozen, the model learns only the classifier. This is simple, stable, and efficient.

When some deeper layers are unfrozen, the model can adapt to the new domain more effectively.


## 5. Fine-Tuning

Fine-tuning means unfreezing selected layers of the pretrained backbone and continuing training with a small learning rate. The goal is to adapt the higher-level feature representations to the new dataset while preserving useful low-level features.

Fine-tuning is typically better than freezing everything when the target dataset is large enough and similar to the source domain.


## 6. Domain Similarity and Domain Mismatch

### Domain Similarity
If the source and target domains are similar, transfer learning works very well. For example, a model trained on natural images can transfer well to other object recognition tasks.

### Domain Mismatch
If the target domain is very different, such as medical images, satellite imagery, or grayscale sketches, the pretrained features may be less useful. In such cases, fine-tuning more layers or training a larger classifier may be necessary.

The closer the target domain is to the source domain, the easier the transfer.


## 7. Workflow 1: Pretrained Model → Freeze Backbone → Replace Classifier → Train

This is the standard feature extraction approach:
- Load a pretrained CNN backbone
- Freeze the convolutional layers
- Replace the original output layer with a new classifier for the target classes
- Train only the new classifier

This is useful when the dataset is small and we want a fast, stable training process.


## 8. Workflow 2: Pretrained Model → Unfreeze Selected Layers → Fine-Tune

This is the common fine-tuning approach:
- Start with a pretrained model
- Replace the classifier head
- Unfreeze some deeper layers
- Train with a smaller learning rate

This allows the model to adapt the higher-level visual features to the new dataset while preserving the earlier general features.

Fine-tuning works well when the target task is similar to the source task and the dataset is not too small.


## 9. Final Interpretation

Transfer learning helps because pretrained networks already encode useful image features. The simplest method is to freeze the backbone and train only a new classifier. If the domain is similar and more accuracy is needed, selected layers can be unfrozen and fine-tuned with a lower learning rate.

The main decision is not just whether to use transfer learning, but also how much of the pretrained model should be kept frozen versus adapted.
