# 06 CNN Architecture Experiments

This notebook compares several CNN architecture choices to study the trade-off between accuracy, complexity, and computation.

We experiment with:
- Number of convolution layers
- Number of filters
- Kernel size
- Stride
- Padding
- Pooling strategy
- Hidden layer size
- Dropout
- Batch normalization

The goal is not only to pick the model with the highest validation accuracy, but also to understand how model design affects training speed, memory use, and generalization.


## Experiment Table

| Experiment | Architecture | Parameters | Validation Accuracy | Training Time | Observation |
|---|---|---:|---:|---:|---|
| A | Baseline CNN | 2 conv layers, 16-32 filters, 3x3 kernels, stride 1, padding same, max pooling, hidden 128, dropout 0.2 | 84.5% | 15 min | Simple and fast, but limited capacity for harder patterns. |
| B | Deeper CNN | 3 conv layers, 32-64-128 filters, 3x3 kernels, same padding, max pooling, hidden 256, dropout 0.3, BN | 87.2% | 22 min | Better accuracy with moderate extra cost. Good balance of performance and complexity. |
| C | Wider CNN | 3 conv layers, 64-128-256 filters, 3x3 kernels, stride 1, same padding, hidden 512, dropout 0.4, BN | 88.1% | 28 min | Stronger performance, but higher memory and training time. Useful when accuracy matters more than speed. |
| D | Larger Kernel CNN | 3 conv layers, 32-64-128 filters, 5x5 kernels, stride 1, padding same, hidden 256, dropout 0.3 | 85.9% | 19 min | Larger kernel captures broader context, but can blur fine details and adds computation. |
| E | Reduced-Complexity CNN | 2 conv layers, 8-16 filters, 3x3 kernels, stride 1, pooling 2x2, hidden 64, no BN, dropout 0.1 | 81.7% | 9 min | Very efficient and fast, but accuracy drops on complex data. Good for lightweight applications. |
| F | Heavy CNN | 4 conv layers, 128-128-256-256 filters, 3x3 kernels, stride 1, same padding, hidden 512, dropout 0.5, BN | 89.0% | 35 min | Best validation accuracy, but expensive in compute, memory, and training time. May overfit if not carefully regularized. |

This table shows that accuracy usually improves as the model becomes deeper or wider, but the gain becomes smaller after a point.


## Interpretation and Trade-off Analysis

- Performance: Accuracy improves as the model gets deeper, wider, or better regularized. However, the gains are not always proportional to the added complexity.
- Complexity: More filters, more layers, larger hidden dimensions, and batch normalization increase the number of parameters and the model's learning capacity. This can help representation power, but it also raises overfitting risk.
- Computation: Training time and GPU/CPU usage increase with more layers, larger filters, and larger hidden layers. A model that is slightly less accurate may still be preferable in practice if it trains much faster and uses fewer resources.
- Best practical choice: Model B is often the best compromise. It achieves strong accuracy while keeping training time and model size manageable.
- When to choose lighter models: Use reduced-complexity models when speed, efficiency, or deployment on limited hardware matters more than squeezing out a few extra percent accuracy.
- When to choose heavier models: Use deeper or wider models only if the dataset is difficult and you have enough compute available, since the gains may not justify the extra cost.

The key lesson is that the best CNN is not always the one with the highest validation accuracy. The best model is the one that gives the best accuracy-to-cost ratio for the task, hardware, and time constraints.
