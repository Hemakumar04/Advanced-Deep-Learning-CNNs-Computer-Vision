## 1. CNN, convolution, filters, and feature maps — conceptual

**Question:** What does a convolutional neural network do with a filter, and what is a feature map?

**Answer:** A CNN slides a small learned filter across an image and computes responses to local patterns. A feature map is the grid of responses produced by applying a filter; high values indicate locations where the learned pattern responds strongly.

## 2. Convolution output size — numerical

**Question:** An input is 32 × 32. A convolution uses a 5 × 5 kernel, padding 2, and stride 1. What is the output height and width? Use `floor((N + 2P - K) / S) + 1`.

**Answer:** `floor((32 + 4 - 5) / 1) + 1 = 32`, so the output is 32 × 32. Padding preserves the spatial size in this case.

## 3. Pooling — numerical

**Question:** Apply 2 × 2 max pooling with stride 2 to `[[1, 3], [4, 2]]`. What value is output?

**Answer:** The output is `[[4]]` because max pooling selects the largest value in the window.

## 4. Padding and stride — comparison

**Question:** What is the difference between padding and stride? What usually happens to feature-map size when stride increases?

**Answer:** Padding adds pixels (commonly zeros) around an input's border. Stride controls how far the filter moves at each step. A larger stride usually makes the output feature map smaller; padding can help preserve border information or spatial size.

## 5. Data augmentation — scenario

**Question:** You have a small training set of centered clothing images. Would a random horizontal flip be a possible augmentation? Which splits should receive it, and what should you check?

**Answer:** It can be a reasonable training augmentation if left-right orientation does not change the label. Apply random augmentation to training images only; keep validation and test preprocessing deterministic. Check that the transformation preserves label meaning for the target domain.

## 6. Batch normalization and dropout — comparison

**Question:** What is one main purpose of batch normalization and one main purpose of dropout? What should happen to dropout during evaluation?

**Answer:** Batch normalization normalizes intermediate activations using batch statistics during training and stored running statistics during evaluation. Dropout randomly disables some activations during training to reduce reliance on particular units. Dropout is disabled during evaluation (`model.eval()`).

## 7. Transfer learning and fine-tuning — scenario

**Question:** You have few labeled images and a pretrained CNN. Describe a common transfer-learning plan and what fine-tuning changes.

**Answer:** Replace the final classifier with one that matches the target classes, freeze most of the pretrained feature extractor, and train the new classifier first. Fine-tuning unfreezes some pretrained layers and updates them with a small learning rate so features adapt to the new task.

## 8. ResNet, MobileNet, and EfficientNet — comparison

**Question:** Match each design idea to the model family: residual/skip connections; mobile/efficient lightweight inference; compound scaling of depth, width, and resolution.

**Answer:** ResNet uses residual (skip) connections. MobileNet is designed for efficient, resource-constrained inference. EfficientNet uses a compound scaling strategy to balance network depth, width, and input resolution.

## 9. Architecture choice — scenario

**Question:** A model must run on a low-power phone. You can choose a large ResNet or a compact MobileNet. What should guide your choice besides accuracy?

**Answer:** Measure latency, memory, model size, energy use, and accuracy on the target phone and representative inputs. MobileNet is a reasonable candidate because it targets efficient inference, but actual hardware measurements and required accuracy should decide.

## 10. Object detection fundamentals — conceptual

**Question:** What information does an object detection prediction usually contain for each object?

**Answer:** A class label, a bounding box describing location (for example `[x_min, y_min, x_max, y_max]`), and a confidence score. An image can have zero, one, or multiple detections.

## 11. Intersection over Union (IoU) — numerical

**Question:** Boxes A=`[0, 0, 10, 10]` and B=`[5, 0, 15, 10]` have intersection area 50 and union area 150. Calculate IoU and interpret it.

**Answer:** `IoU = 50 / 150 = 1/3 ≈ 0.33`. The boxes overlap, but cover substantial non-shared area. IoU ranges from 0 (no overlap) to 1 (identical boxes).

## 12. Non-Maximum Suppression (NMS) — scenario

**Question:** A detector predicts three boxes for one class. Two boxes overlap above the NMS threshold; the first has confidence 0.91 and the second 0.74. What does NMS generally do?

**Answer:** Keep the higher-confidence box and suppress the overlapping lower-confidence box, then compare remaining candidates. NMS reduces duplicate detections; it does not improve the detector's underlying classification or box regression.

## 13. Semantic, instance, and panoptic segmentation — comparison

**Question:** An image contains two cars and a road. How would semantic, instance, and panoptic segmentation represent them?

**Answer:** Semantic segmentation labels both cars with the same `car` class and road pixels as `road`. Instance segmentation gives each car a separate object identity and mask. Panoptic segmentation combines class labels for all pixels with separate identities for individual countable objects such as the two cars.

## 14. Bounding boxes and masks — comparison

**Question:** When is a mask more useful than a bounding box?

**Answer:** When the precise object outline or pixel area matters, such as measuring a lesion, separating overlapping objects, or estimating an object's shape. A bounding box is simpler and often sufficient when only approximate location is needed.

## 15. Model optimization — scenario

**Question:** Model A is 80 MB and 2 ms per image at 91% accuracy. Model B is 12 MB and 7 ms per image at 89% accuracy. A device allows at most 20 MB and requires at least 88% accuracy. Which model meets the stated constraints?

**Answer:** Model B meets the size and accuracy constraints; Model A exceeds the size limit. The slower speed of B may still matter, so verify that 7 ms also meets the application's latency requirement.

## 16. Error analysis — debugging a confusion matrix

**Question:** A classifier has high overall accuracy, but the confusion matrix shows many shirts predicted as coats. What should you investigate before changing the architecture?

**Answer:** Inspect misclassified images and labels for ambiguity or label errors; compare per-class recall and sample counts; check preprocessing and train/validation distribution; and determine whether shirts versus coats are genuinely visually ambiguous at the image resolution. Then consider representative data, label guidance, or targeted training changes.

## 17. Debugging a CNN dimension error — debugging exercise

**Question:** Input shape is `(batch, 1, 28, 28)`. The network applies `Conv2d(1, 8, kernel_size=3)` with default padding/stride, then `MaxPool2d(2)`, then `Conv2d(8, 16, kernel_size=3)`, then `MaxPool2d(2)`. A fully connected layer expects `16 * 7 * 7` values. Why can the linear layer fail, and what is the actual flattened size?

**Answer:** With no padding, the spatial sizes are `28 → 26 → 13 → 11 → 5`. The actual flattened size is `16 * 5 * 5 = 400`, not `16 * 7 * 7`. Update the linear input size to 400 or use adaptive pooling to avoid hard-coding the spatial dimensions.

## 18. Coding exercise: compute convolution output size

**Question:** Complete the calculation for a 28-pixel input, 3-pixel kernel, padding 1, and stride 2. What is the output size?

**Answer:** Use `floor((N + 2P - K) / S) + 1`; the result is 14.

```python
import math
input_size, kernel, padding, stride = 28, 3, 1, 2
output_size = math.floor((input_size + 2 * padding - kernel) / stride) + 1
print(output_size)
assert output_size == 14
```

## 19. Coding exercise: calculate IoU

**Question:** Write a small function that computes IoU for two `[x_min, y_min, x_max, y_max]` boxes. What should it return for `[0, 0, 10, 10]` and `[5, 0, 15, 10]`?

**Answer:** It should return about `0.333`.

```python
def box_iou(box_a, box_b):
    left = max(box_a[0], box_b[0])
    top = max(box_a[1], box_b[1])
    right = min(box_a[2], box_b[2])
    bottom = min(box_a[3], box_b[3])
    intersection = max(0, right - left) * max(0, bottom - top)
    area_a = (box_a[2] - box_a[0]) * (box_a[3] - box_a[1])
    area_b = (box_b[2] - box_b[0]) * (box_b[3] - box_b[1])
    return intersection / (area_a + area_b - intersection)

result = box_iou([0, 0, 10, 10], [5, 0, 15, 10])
print(round(result, 3))
assert round(result, 3) == 0.333
```

## 20. Coding exercise: inspect feature-map shape

**Question:** Use a convolution with 4 output filters, kernel size 3, padding 1, and default stride on a batch shaped `(2, 1, 28, 28)`. What feature-map shape should be printed?

**Answer:** `(2, 4, 28, 28)`: the batch size stays 2, there are 4 output channels, and padding preserves height and width.

```python
import torch
from torch import nn
conv = nn.Conv2d(in_channels=1, out_channels=4, kernel_size=3, padding=1)
images = torch.zeros(2, 1, 28, 28)
feature_maps = conv(images)
print(tuple(feature_maps.shape))
assert tuple(feature_maps.shape) == (2, 4, 28, 28)
```

## 21. Coding/debugging exercise: correct train and evaluation modes

**Question:** A developer forgets to call `model.eval()` before evaluating a network with dropout. What is wrong, and how should the model be prepared for prediction?

**Answer:** Dropout remains active in training mode, making predictions stochastic and changing behavior from deployment. Call `model.eval()` and use `torch.inference_mode()` for prediction.

```python
model = nn.Sequential(nn.Flatten(), nn.Linear(4, 4), nn.Dropout(p=0.5), nn.Linear(4, 2))
model.eval()
with torch.inference_mode():
    prediction_scores = model(torch.ones(1, 1, 2, 2))
print("Evaluation mode:", not model.training)
print("Output shape:", tuple(prediction_scores.shape))
assert not model.training and tuple(prediction_scores.shape) == (1, 2)
```

## 22. Coding exercise: simple NMS decision

**Question:** Given confidence scores `[0.9, 0.7]`, box IoU `0.8`, and NMS threshold `0.5`, which prediction is kept if both boxes represent the same class and object?

**Answer:** Keep the first prediction (score 0.9) and suppress the second because their IoU exceeds 0.5.

```python
scores = [0.9, 0.7]
overlap_iou = 0.8
threshold = 0.5
kept = [0, 1] if overlap_iou <= threshold else [max(range(len(scores)), key=scores.__getitem__)]
print("Kept prediction index:", kept[0])
assert kept == [0]
```

## 23. Debugging a misleading accuracy result — scenario

**Question:** A validation score is excellent, but test accuracy is much lower. Name two checks you should make before claiming the model overfits.

**Answer:** Check that validation and test preprocessing, class-index mappings, and evaluation mode match; verify that the splits are representative and have no leakage or distribution shift; inspect per-class metrics and labels. Overfitting is possible, but these other causes should be ruled out.
