# 07 Batch Normalization and Dropout

## 1. Key Concepts

### Internal Distribution Changes
During training, the distribution of activations keeps changing because network parameters are updated every batch. This is often called internal covariate shift. It makes optimization harder, especially in deep CNNs.

### Batch Normalization
Batch normalization normalizes each feature across the current mini-batch. This keeps activations stable and allows the network to train faster and more reliably. In practice, BatchNorm usually improves convergence speed and can lead to better final accuracy.

### Normalization vs Standardization
- Normalization: scales values into a fixed range, such as 0 to 1.
- Standardization: transforms data to zero mean and unit variance.
BatchNorm applies a form of standardization to activations inside the network, not just to the raw input data.

### Dropout
Dropout randomly disables neurons during training. This prevents co-adaptation and reduces overfitting. However, it can slow training because the model learns with a different network structure each step.

### Training vs Evaluation Mode
In PyTorch, the model behaves differently in training and evaluation mode:
- `model.train()` enables training behavior such as dropout and BatchNorm statistics updates.
- `model.eval()` disables dropout and uses the accumulated running statistics of BatchNorm.

This difference is critical because evaluation must reflect the real deployed model, not the stochastic training-time version.


```python
import torch
import torch.nn as nn
x = torch.randn(32, 64, 28, 28)
conv = nn.Conv2d(64, 64, kernel_size=3, padding=1)
bn = nn.BatchNorm2d(64)
drop = nn.Dropout(p=0.3)
y = conv(x)
y = bn(y)
y = drop(y)
print(y.shape)
```

## 3. Analysis of Results

- Baseline CNN shows the weakest stability because activations shift during training, making optimization harder.
- BatchNorm improves optimization by keeping activations normalized within each mini-batch, which usually leads to higher accuracy and faster convergence.
- Dropout improves generalization by preventing the model from relying too much on any single neuron, but it can make training noisier and slightly slower.
- The combination of BatchNorm + Dropout is usually the most effective because BatchNorm stabilizes learning while Dropout reduces overfitting.
- The best model is not always the most complex one. A simple CNN may be preferred for speed and simplicity, while BatchNorm + Dropout is preferred when generalization matters more.

## 4. Final Interpretation

Batch normalization and dropout are complementary techniques. BatchNorm improves the internal flow of information in the network, while dropout regularizes the model by preventing over-reliance on specific features. Together, they often give the strongest performance, but the final choice depends on the trade-off between accuracy, training stability, and computational cost.
